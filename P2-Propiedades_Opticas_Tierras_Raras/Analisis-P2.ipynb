```python
import numpy as np
import matplotlib.pyplot as plt
import csv
import math

import pandas as pd
```

$$ E = \frac{10^7}{\lambda}$$

La $E$ está en $cm^{-1}$ y la lambda en $nm$.

```python
# Definimos una función para calcular la energía

def energia(longitud_onda):
    E = (1e7)/longitud_onda
    return E
```

## **Medición 1**

**Notación:**

dy: disprocio \
med: medición

```python
# Leemos el archivo de datos
dy_med_1 = pd.read_csv("EuVLx#11.sp",sep='\t',skiprows=54,header=None,dtype='float')

# Le ponemos nombre a las columnas
dy_med_1.columns=(["x","y"])

# Imprimimos la tabla de datos
dy_med_1
```

```python
# Guardamos las columnas en una lista cada una para poder graficarlas
dy_med_1_x = dy_med_1["x"].values
dy_med_1_y = dy_med_1["y"].values

# Las visualizamos
#print(dy_med_1_x)
#print(dy_med_1_y)
```

**Hacemos una gráfica inicial de los datos**

```python
plt.plot(dy_med_1_x,dy_med_1_y,'.-',c='dodgerblue')
plt.title('Dy - medición 1')
plt.xlabel(r'$\lambda$ [nm]'), plt.ylabel(r'Intensidad')
plt.grid()
```

**Calculamos cada pico**

```python
# Ubicamos el pico (1) más grande
dy_med_1_maxInt_1 = max(dy_med_1_y)
dy_med_1_pico_1 = dy_med_1_x[np.argmax(dy_med_1_y)]

print('Intensidad max 1 =',dy_med_1_maxInt_1)
print(r'Lambda max 1 =',dy_med_1_pico_1)

# Ubicamos los demás picos ajustando el rango a mano (viendo la gráfica)
# pico (2)
dy_med_1_maxInt_2 = max(dy_med_1_y[0:100])
dy_med_1_pico_2 = dy_med_1_x[np.argmax(dy_med_1_y[0:100])]

print('Intensidad max 2 =',dy_med_1_maxInt_2)
print(r'Lambda max 2 =',dy_med_1_pico_2)

# pico (3)
dy_med_1_maxInt_3 = max(dy_med_1_y[400:500])
dy_med_1_pico_3 = dy_med_1_x[400 + np.argmax(dy_med_1_y[400:500])]

print('Intensidad max 3 =',dy_med_1_maxInt_3)
print(r'Lambda max 3 =',dy_med_1_pico_3)

# pico (4)
dy_med_1_maxInt_4 = max(dy_med_1_y[600:700])
dy_med_1_pico_4 = dy_med_1_x[600 + np.argmax(dy_med_1_y[600:700])]

print('Intensidad max 4 =',dy_med_1_maxInt_4)
print(r'Lambda max 4 =',dy_med_1_pico_4)
```

**Hacemos otra gráfica localizando estos puntos máximos para confirmar**

```python
plt.plot(dy_med_1_x,dy_med_1_y,'.',c='dodgerblue')
plt.title('Dy - medición 1')
plt.xlabel(r'$\lambda$ [nm]'), plt.ylabel(r'Intensidad')
plt.axvline(dy_med_1_pico_1,c='r'), plt.axhline(dy_med_1_maxInt_1,c='r') # pico (1)
plt.axvline(dy_med_1_pico_2,c='g'), plt.axhline(dy_med_1_maxInt_2,c='g') # pico (2)
plt.axvline(dy_med_1_pico_3,c='y'), plt.axhline(dy_med_1_maxInt_3,c='y') # pico (3)
plt.axvline(dy_med_1_pico_4,c='m'), plt.axhline(dy_med_1_maxInt_4,c='m') # pico (4)
#plt.grid()
```

**Ya verificados, recordamos los picos y calculamos la energía correspondiente**

```python
# pico (1)
dy_med_1_ener_1 = energia(dy_med_1_pico_1)

print(r'Lambda max 1 =',dy_med_1_pico_1)
print(f"Enegía correspondiente = {dy_med_1_ener_1:.2f}")

# pico (2)
dy_med_1_ener_2 = energia(dy_med_1_pico_2)

print(r'Lambda max 2 =',dy_med_1_pico_2)
print(f"Enegía correspondiente = {dy_med_1_ener_2:.2f}")

# pico (3)
dy_med_1_ener_3 = energia(dy_med_1_pico_3)

print(r'Lambda max 3 =',dy_med_1_pico_3)
print(f"Enegía correspondiente = {dy_med_1_ener_3:.2f}")

# pico (4)
dy_med_1_ener_4 = energia(dy_med_1_pico_4)

print(r'Lambda max 4 =',dy_med_1_pico_4)
print(f"Enegía correspondiente = {dy_med_1_ener_4:.2f}")
```
